## 1.1 Vectors

A vector is just an ordered list of numbers. The most direct way to hold one in Python is a plain list: numbers inside square brackets, separated by commas. `len()` tells you how many elements it has — that number is the *dimension* of the vector.

```python
x = [-1.1, 0.0, 3.6, -7.2]
len(x)
```

Watch out: parentheses give you a **tuple**, not a vector. A tuple is a fixed group of values, and it does not behave like a vector for arithmetic.

```python
y = (1, 2)
y
```

For real work we use a **numpy array** instead of a list. Arrays understand vector arithmetic (adding, scaling, dot products), which lists do not. Import numpy once, then build the array from a list.

```python
import numpy as np
x = np.array([-1.1, 0.0, 3.6, -7.2])
len(x)
```

### Indexing

Pick out one element with `x[i]`. Python counts from **0**, so a vector of length $n$ has indices `0` through `n-1`. Here `x[2]` is the third entry.

```python
import numpy as np
x = np.array([-1.1, 0.0, 3.6, -7.2])
x[2]
```

Put an index on the left of `=` to change that entry in place.

```python
x[2] = 20.0
print(x)
```

A handy shortcut: `-1` means the **last** element (`-2` the second-to-last, and so on).

```python
x[-1]
```

### Assignment versus copying

This surprises people coming from MATLAB. Writing `y = x` does **not** copy the array — it just gives the same array a second name. Change `x` and `y` changes too, because they point at the same data.

```python
import numpy as np
x = np.array([-1.1, 0.0, 3.6, -7.2])
y = x
x[2] = 20.0
print(y)
```

To get an independent copy, use `.copy()`. Now editing `x` leaves `y` untouched.

```python
import numpy as np
x = np.array([-1.1, 0.0, 3.6, -7.2])
y = x.copy()
x[2] = 20.0
print(y)
```

### Vector equality

On numpy arrays, `==` compares **element by element** and returns an array of True/False, one per entry.

```python
import numpy as np
x = np.array([-1.1, 0.0, 3.6, -7.2])
y = x.copy()
x == y
```

```python
import numpy as np
x = np.array([-1.1, 0.0, 3.6, -7.2])
y = x.copy()
y[3] = 9.0
x == y
```

Plain lists behave differently: `==` on two lists gives a **single** True only when they have the same length and identical entries.

```python
x = [-1.1, 0.0, 3.6, -7.2]
y = x.copy()
x == y
```

```python
x = [-1.1, 0.0, 3.6, -7.2]
y = x.copy()
y[3] = 9.0
x == y
```

### Scalars versus 1-vectors

In math a 1-vector is basically a scalar. In Python they are not the same object: a one-element list `[2.4]` is not the number `2.4`.

```python
x = 2.4
y = [2.4]
x == y
```

```python
y[0] == 2.4
```

A one-element numpy array does compare like a scalar (element-wise), so it feels more like the math.

```python
import numpy as np
x = 2.4
y = np.array([2.4])
x == y
```

### Block and stacked vectors

To stack vectors end-to-end into one longer vector, use `np.concatenate()`. Note the **double parentheses**: the vectors go inside one tuple.

```python
import numpy as np
x = np.array([1, -2])
y = np.array([1, 1, 0])
z = np.concatenate((x, y))
print(z)
```

### Subvectors and slicing

`x[a:b]` grabs a *slice* — entries from index `a` up to but **not including** `b`. So `x[1:4]` gives the 2nd, 3rd, and 4th entries.

```python
import numpy as np
x = np.array([1, 8, 3, 2, 1, 9, 7])
y = x[1:4]
print(y)
```

You can also assign into a slice to overwrite several entries at once.

```python
x[3:6] = [100, 200, 300]
print(x)
```

Leaving a side of the colon blank means "to the end" or "from the start". `x[2:]` is everything from index 2 on; `x[:-1]` is everything except the last.

```python
import numpy as np
x = np.array([1, 8, 3, 2, 1, 9, 7])
x[2:]
```

```python
x[:-1]
```

A third number is the **step**. `x[1:4:2]` takes every 2nd element in that range, and `x[::-1]` reverses the whole vector.

```python
import numpy as np
x = np.array([1, 8, 3, 2, 1, 9, 7])
print(x[1:4:2])
print(x[::-1])
```

### Vector of first differences

Slicing makes it easy to compute successive differences $d_i = x_{i+1} - x_i$: line up the vector with a shifted copy of itself and subtract.

```python
import numpy as np
x = np.array([1, 8, 3, 2, 1, 9, 7])
d = x[1:] - x[:-1]
print(d)
```

### Lists of vectors

Sometimes you want a collection of vectors. Put them in a list; then `L[i]` is the i-th vector and `L[i][j]` is the j-th entry of that vector.

```python
x = [1, 0]
y = [1, -1]
z = [0, 1]
list_of_vectors = [x, y, z]
list_of_vectors[1]  # second vector
```

```python
list_of_vectors[1][0]  # first entry of the second vector
```

## 1.2 Vector addition

For arrays of the same size, `+` and `-` add and subtract entry by entry — exactly the math definition.

```python
import numpy as np
x = np.array([1, 2, 3])
y = np.array([100, 200, 300])
print('Sum:', x + y)
print('Difference:', x - y)
```

## 1.3 Scalar-vector multiplication

Multiplying a vector by a number scales every entry. Order does not matter (`a*x` or `x*a`), and `/` divides every entry.

```python
import numpy as np
x = np.array([1, 2, 3])
print(2.2 * x)
```

```python
import numpy as np
x = np.array([1, 2, 3])
print(x / 2.2)
```

### Scalar-vector addition

Python also lets you add a number to an array, which adds it to every entry. Convenient, but note this is **not** standard math notation (mathematically you would add a multiple of the ones vector).

```python
import numpy as np
x = np.array([1, 2, 3, 4])
print(x + 2)
```

### Elementwise operations

`*`, `/`, `**` between two arrays act entry by entry. Example: asset returns are the elementwise change in price divided by the starting price.

```python
import numpy as np
p_initial = np.array([22.15, 89.32, 56.77])
p_final = np.array([23.05, 87.32, 53.13])
r = (p_final - p_initial) / p_initial
r
```

### Linear combination

Combining scaling and addition gives a *linear combination* $\alpha a + \beta b$ — a core operation in linear algebra.

```python
import numpy as np
a = np.array([1, 2])
b = np.array([3, 4])
alpha = -0.5
beta = 1.5
c = alpha * a + beta * b
print(c)
```

We can wrap it in a function that takes a list of coefficients and a list of vectors and returns their linear combination.

```python
def lincomb(coef, vectors):
    n = len(vectors[0])
    comb = np.zeros(n)
    for i in range(len(vectors)):
        comb = comb + coef[i] * vectors[i]
    return comb

lincomb([alpha, beta], [a, b])
```

A shorter one-line version does the same thing.

```python
def lincomb(coef, vectors):
    return sum(coef[i] * vectors[i] for i in range(len(vectors)))
```

### Checking properties

A good habit: test an identity numerically with random inputs. Here we confirm the distributive law $\beta(a+b) = \beta a + \beta b$. The two sides should match (up to tiny floating-point error).

```python
import numpy as np
a = np.random.random(3)
b = np.random.random(3)
beta = np.random.random()
lhs = beta * (a + b)
rhs = beta * a + beta * b
print('LHS:', lhs)
print('RHS:', rhs)
```

## 1.4 Inner product

The inner (dot) product $x^T y$ multiplies matching entries and sums them, giving a single number. Use `np.inner(x, y)`.

```python
import numpy as np
x = np.array([-1, 2, 2])
y = np.array([1, 0, -3])
print(np.inner(x, y))
```

The `@` operator does the same thing and is often the cleanest to read.

```python
import numpy as np
x = np.array([-1, 2, 2])
y = np.array([1, 0, -3])
x @ y
```

### Net present value

A classic inner-product use: discount a cash-flow vector by $(1+r)^{-i}$ and sum to get the net present value.

```python
import numpy as np
c = np.array([0.1, 0.1, 0.1, 1.1])  # cash flows
n = len(c)
r = 0.05  # 5% per-period interest
d = np.array([(1 + r) ** -i for i in range(n)])  # discount factors
NPV = c @ d
print(NPV)
```

### Total school-age population

An inner product with a **selector** vector (ones where you want to count, zeros elsewhere) picks out and sums a subset of entries.

```python
import numpy as np
x = np.random.random(100)  # ages 0..99
s = np.concatenate([np.zeros(5), np.ones(14), np.zeros(81)])  # ages 5..18
school_age_pop = s @ x
school_age_pop
```

## 1.5 Complexity of vector computations

Computers use finite-precision arithmetic, so algebraically equal expressions can differ by a minuscule amount. Here $(a+b)(a-b)$ vs $a^2-b^2$ differ only by round-off.

```python
import numpy as np
a = np.random.random()
b = np.random.random()
print((a + b) * (a - b) - (a**2 - b**2))
```

The cost of an inner product grows linearly with the vector length ($2n-1$ flops). We can time it — and it often runs faster the second time thanks to caching.

```python
import numpy as np
import time
a = np.random.random(10**5)
b = np.random.random(10**5)
start = time.time()
a @ b
end = time.time()
print(end - start)
```

```python
start = time.time()
a @ b
end = time.time()
print(end - start)
```

### Sparse vectors

When most entries are zero it wastes memory to store them all. A **sparse** vector keeps only the nonzero entries (with their positions). `scipy.sparse` handles this.

```python
import numpy as np
from scipy import sparse
I = np.array([4, 7, 8, 9])   # positions of nonzeros
J = np.array([0, 0, 0, 0])
V = np.array([100, 200, 300, 400])  # values
A = sparse.coo_matrix((V, (I, J)), shape=(10000, 1))
A
```
