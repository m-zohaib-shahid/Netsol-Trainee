# Lecture 1 — NumPy

NumPy provides the fast, memory-efficient **ndarray** and vectorized operations that power the entire Python data stack.

````python
import numpy as np
print('NumPy', np.__version__)
````

## 1.1 Creating Arrays

````python
# From Python lists
a = np.array([1, 2, 3, 4, 5])
m = np.array([[1, 2, 3], [4, 5, 6]])
print("1D:", a)
print("2D:\n", m)
````

````python
# Generators
print("zeros:\n", np.zeros((2, 3)))
print("ones:\n", np.ones((2, 2)))
print("full:\n", np.full((2, 2), 7))
print("arange:", np.arange(0, 10, 2))
print("linspace:", np.linspace(0, 1, 5))
print("eye:\n", np.eye(3))
````

````python
# dtype control + type promotion edge cases
print(np.array([1, 2, 3], dtype=float))
print(np.array([1, 2, 3.5]).dtype)          # promoted to float64
print(np.array([1, 2, "x"]).dtype)          # everything becomes string '<U...'
````

### Array attributes

````python
m = np.arange(6).reshape(2, 3)
print("shape:", m.shape, "| ndim:", m.ndim, "| size:", m.size, "| dtype:", m.dtype, "| itemsize:", m.itemsize)
````

### Reshaping, flattening, stacking

````python
a = np.arange(12)
print("reshape 3x4:\n", a.reshape(3, 4))
print("auto dim (-1):\n", a.reshape(2, -1))     # NumPy infers the 6
print("transpose:\n", a.reshape(3, 4).T)
print("ravel:", a.reshape(3, 4).ravel())
````

````python
# Combining arrays
x = np.array([1, 2, 3]); y = np.array([4, 5, 6])
print("concatenate:", np.concatenate([x, y]))
print("vstack:\n", np.vstack([x, y]))
print("hstack:", np.hstack([x, y]))
print("column_stack:\n", np.column_stack([x, y]))
````

### Useful helpers: sort, unique

````python
u = np.array([3, 1, 2, 3, 1, 5])
print("sorted:", np.sort(u))
print("unique:", np.unique(u))
vals, counts = np.unique(u, return_counts=True)
print("value counts:", dict(zip(vals, counts)))
````

⚠️ **Common Mistake 1.** `reshape` needs the sizes to match. `np.arange(10).reshape(3,4)` fails because 10 ≠ 12. Also, `a.reshape(...)` returns a **new view/array** — it does *not* modify `a` in place.

````python
a = np.arange(6)
a.reshape(2, 3)          # result discarded!
print("a is unchanged:", a.shape)
a = a.reshape(2, 3)      # correct: reassign
print("now:", a.shape)
````

## 1.2 Indexing & Slicing

````python
a = np.array([10, 20, 30, 40, 50])
print(a[0], a[-1], a[1:4], a[::2], a[::-1])
````

````python
m = np.arange(1, 10).reshape(3, 3)
print("m:\n", m)
print("(1,2):", m[1, 2])
print("row 0:", m[0])
print("col 1:", m[:, 1])
print("2x2 block:\n", m[:2, :2])
````

### Boolean / mask indexing & `np.where`

````python
a = np.array([5, 12, 7, 20, 3, 18])
print("mask:", a > 10)
print("values > 10:", a[a > 10])
print("where clip:", np.where(a > 10, a, 0))
print("count > 10:", np.sum(a > 10))
print("any/all > 10:", np.any(a > 10), np.all(a > 10))
````

### Fancy indexing

````python
a = np.array([100, 200, 300, 400, 500])
print(a[[0, 2, 4]])
m = np.arange(1, 10).reshape(3, 3)
print("pick (0,0),(1,1),(2,2):", m[[0, 1, 2], [0, 1, 2]])   # diagonal
````

⚠️ **Common Mistake 2.** A **slice is a view**, not a copy — writing to it changes the original. Use `.copy()` when you need independence.

````python
a = np.array([1, 2, 3, 4, 5])
s = a[1:4]
s[0] = 999
print("original changed!:", a)      # a[1] is now 999

b = np.array([1, 2, 3, 4, 5])
s2 = b[1:4].copy()
s2[0] = 999
print("original safe:", b)
````

## 1.3 Broadcasting

Dimensions are compatible when equal or one is **1**. NumPy stretches the smaller array.

````python
a = np.array([1, 2, 3, 4])
print("+ scalar:", a + 10)
print("* 2:", a * 2)
````

````python
col = np.array([[0], [10], [20]])   # (3,1)
row = np.array([1, 2, 3])           # (3,)
print("col+row grid:\n", col + row)
````

````python
# Center each column by its mean
data = np.array([[10, 20, 30], [40, 50, 60]], dtype=float)
print("centered:\n", data - data.mean(axis=0))
````

⚠️ **Common Mistake 3.** Shape mismatches that don't broadcast raise errors. A (3,) array can't add to a (2,) array. Reshape to make one dimension 1 when you *intend* an outer combination.

````python
try:
    np.array([1, 2, 3]) + np.array([1, 2])
except ValueError as e:
    print("ValueError:", e)
````

## 1.4 Operations & Aggregations

````python
a = np.array([1, 2, 3]); b = np.array([10, 20, 30])
print("add:", a + b, "| mul:", a * b)
print("sqrt:", np.sqrt(b), "| exp:", np.round(np.exp(a), 2))
print("dot product:", np.dot(a, b))
````

````python
m = np.array([[1, 2, 3], [4, 5, 6]])
print("sum:", m.sum(), "| mean:", m.mean(), "| std:", round(m.std(), 3))
print("sum axis=0 (cols):", m.sum(axis=0))
print("sum axis=1 (rows):", m.sum(axis=1))
print("argmax:", m.argmax(), "| argmin:", m.argmin())
print("cumsum:", m.cumsum())
````

### Random numbers (with a seed for reproducibility)

````python
rng = np.random.default_rng(42)
print("uniform:", np.round(rng.random(3), 3))
print("integers:", rng.integers(1, 7, size=5))
print("normal:", np.round(rng.normal(0, 1, 4), 3))
````

⚠️ **Common Mistake 4.** `axis` is the dimension that gets **collapsed**. `axis=0` collapses rows → gives one value **per column**. Beginners often swap these.

# Lecture 2 — Pandas I

Pandas adds labelled, table-like structures on top of NumPy: the 1D **Series** and 2D **DataFrame** — the workhorses of data cleaning and analysis.

````python
import pandas as pd
import numpy as np
print('pandas', pd.__version__)
````

## 2.1 Series & DataFrame

````python
s = pd.Series([10, 20, 30, 40])
print(s)
s2 = pd.Series([85, 90, 78], index=["Ali", "Sara", "Zed"])
print("\nSara:", s2["Sara"])
print("vectorized:", (s2 + 5).tolist())
````

````python
df = pd.DataFrame({
    "name":  ["Ali", "Sara", "Zed", "Mina", "Omar"],
    "age":   [23, 35, 29, 41, 30],
    "city":  ["Lahore", "Karachi", "Lahore", "Quetta", "Lahore"],
    "score": [85, 90, 78, 88, 72],
})
df
````

### Inspecting

````python
print("shape:", df.shape)
print("columns:", list(df.columns))
print("dtypes:\n", df.dtypes)
df.head(2)
````

````python
df.describe()
````

````python
df.info()
````

### Quick summaries: `value_counts`, `unique`, `sort_values`

````python
print(df["city"].value_counts())
print("\nunique cities:", df["city"].unique())
print("\nsorted by score:")
df.sort_values("score", ascending=False)
````

## 2.2 Selection — columns, `.loc`, `.iloc`

````python
print(df["name"])            # Series
df[["name", "score"]]        # DataFrame
````

````python
print("loc row 0:\n", df.loc[0])
print("\niloc row 0:\n", df.iloc[0])
print("\nloc rows 0-2, cols:\n", df.loc[0:2, ["name", "score"]])
print("\niloc block:\n", df.iloc[0:2, 0:2])
````

### Boolean filtering (single & combined conditions)

````python
print(df[df["age"] > 30])
print("\nLahore & score>80:")
print(df[(df["city"] == "Lahore") & (df["score"] > 80)])
print("\nisin:")
print(df[df["city"].isin(["Karachi", "Quetta"])])
````

⚠️ **Common Mistake 5.** Combine conditions with `&` / `|` (not `and`/`or`) and **wrap each in parentheses** — operator precedence will bite you otherwise.

````python
# Wrong: df[df["age"] > 30 & df["score"] > 80]  -> error / wrong result
# Right:
print(df[(df["age"] > 30) & (df["score"] > 80)])
````

### `set_index` / `reset_index`

````python
di = df.set_index("name")
print(di.loc["Sara"])
di.reset_index().head()
````

## 2.3 Cleaning

````python
messy = pd.DataFrame({
    "Full Name": [" ali ", "SARA", "Zed", "Zed", "mina"],
    "Age":       ["23", "35", "29", "29", "41"],
    "City":      ["lahore", "karachi", "lahore", "lahore", "quetta"],
})
messy
````

````python
messy = messy.rename(columns={"Full Name": "name", "Age": "age", "City": "city"})
messy["age"] = messy["age"].astype(int)                 # text -> int
messy["name"] = messy["name"].str.strip().str.title()   # tidy strings
messy["city"] = messy["city"].str.title()
messy = messy.drop_duplicates()
messy
````

### `apply`, `map`, and binning with `cut`

````python
df["grade"] = df["score"].apply(lambda x: "A" if x >= 85 else "B")
df["age_group"] = pd.cut(df["age"], bins=[0, 30, 40, 100],
                         labels=["<=30", "31-40", "40+"])
df[["name", "score", "grade", "age", "age_group"]]
````

⚠️ **Common Mistake 6.** Chained assignment like `df[df.a>0]['b'] = 1` triggers a `SettingWithCopyWarning` and may not work. Use `.loc` for assignment: `df.loc[df.a>0, 'b'] = 1`.

````python
df.loc[df["score"] < 75, "grade"] = "C"
df[["name", "score", "grade"]]
````

## 2.4 Missing Data

````python
nd = pd.DataFrame({
    "name":  ["Ali", "Sara", "Zed", "Mina"],
    "age":   [23, np.nan, 29, np.nan],
    "score": [85, 90, np.nan, 88],
})
nd
````

````python
print(nd.isnull())
print("\nper column:\n", nd.isnull().sum())
print("\nany missing?:", nd.isnull().values.any())
````

````python
print("dropna rows:\n", nd.dropna())
print("\ndropna cols:\n", nd.dropna(axis=1))
print("\nthresh=2 (keep rows with >=2 non-null):\n", nd.dropna(thresh=2))
````

````python
filled = nd.copy()
filled["age"] = filled["age"].fillna(filled["age"].mean())
filled["score"] = filled["score"].fillna(filled["score"].median())
filled
````

⚠️ **Common Mistake 7.** Most pandas methods return a **new** object and don't modify in place. `df.fillna(0)` alone does nothing lasting — reassign (`df = df.fillna(0)`) or assign the column back.
